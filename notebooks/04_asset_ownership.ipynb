```python
import pandas as pd
import altair as alt
from pathlib import Path
```

```python
# Project directories
DATA_DIR = Path("../data")
IMAGES_DIR = Path("../images")

# Excel file
file_path = DATA_DIR / "eic2025_genero.xlsx"
```

```python
df_22 = pd.read_excel(
    file_path,
    sheet_name="22",
    header=None
)

print("Shape:", df_22.shape)

display(df_22.iloc[:12, :20])
```

```python
display(df_22.iloc[8:16, [0, 1, 13, 14, 15, 16]])
```

```python
cars = df_22.loc[
    df_22[1] == "Valor",
    [0, 13, 14, 15, 16]
].copy()

cars.columns = [
    "state", "total", "women", "men", "gap"
]

cars.head()
```

```python
# Clean state names
cars["state"] = cars["state"].str.replace(
    r"^\d{2}\s+", "", regex=True
)

# Separate national reference
national = cars.loc[
    cars["state"] == "Estados Unidos Mexicanos"
].copy()

# Keep only states
cars_states = cars.loc[
    cars["state"] != "Estados Unidos Mexicanos"
].copy()

print("Number of states:", len(cars_states))
print("National gap:", national["gap"].iloc[0])
```

```python
# Top 10 states with the largest gender gaps
top10_cars = (
    cars_states
    .sort_values("gap")
    .head(10)
    .copy()
)

top10_cars[["state", "women", "men", "gap"]].round(2)
```

```python
df_home = pd.read_excel(
    file_path,
    sheet_name="24",
    header=None
)

df_home.iloc[8:12, :9]
```

```python
ownership_data = pd.DataFrame({
    "indicator": [
        "Vehicle availability",
        "Vehicle availability",
        "Sole home ownership",
        "Sole home ownership"
    ],
    "gender": ["Women", "Men", "Women", "Men"],
    "percentage": [
        national["women"].iloc[0],
        national["men"].iloc[0],
        df_home.loc[9, 3],
        df_home.loc[9, 4]
    ]
})

ownership_data
```

```python
# Base chart
base = alt.Chart(ownership_data).encode(
    x=alt.X(
        "gender:N",
        sort=["Women", "Men"],
        title=None,
        axis=alt.Axis(labelAngle=0, labelFontSize=12)
    ),
    y=alt.Y(
        "percentage:Q",
        title="Percentage (%)",
        scale=alt.Scale(domain=[0, 65]),
        axis=alt.Axis(grid=False)
    ),
    color=alt.Color(
        "gender:N",
        scale=alt.Scale(
            domain=["Women", "Men"],
            range=["#8624f5", "#1fc3aa"]
        ),
        legend=None
    )
)

# Bars
bars = base.mark_bar(size=65)

# Percentage labels
labels = base.mark_text(
    dy=-10,
    fontSize=13,
    fontWeight="bold",
    color="#333333"
).encode(
    text=alt.Text("percentage:Q", format=".1f")
)

# Combine charts
ownership_chart = (
    (bars + labels)
    .properties(width=240, height=300)
    .facet(
        column=alt.Column(
            "indicator:N",
            sort=[
                "Vehicle availability",
                "Sole home ownership"
            ],
            title=None,
            header=alt.Header(labelFontSize=15)
        )
    )
    .properties(
        title=alt.TitleParams(
            text="Gender disparities in household assets",
            subtitle="Mexico, 2025 | National estimates",
            fontSize=20,
            subtitleFontSize=13,
            anchor="middle"
        )
    )
    .configure_view(stroke=None)
)

ownership_chart
```

```python
ownership_chart.save("../images/ownership.png", scale_factor=2)
```
