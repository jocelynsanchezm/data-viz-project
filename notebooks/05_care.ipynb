```python
import pandas as pd
import altair as alt
from pathlib import Path
import numpy as np
```

```python
file_path = "../data/care/cuidados_estimadores.xlsx"

xls = pd.ExcelFile(file_path)

xls.sheet_names
```

```python
df_care = pd.read_excel(
    file_path,
    sheet_name="Cuadro 3.4",
    header=None
)

df_care.iloc[5:16, :]
```

```python
# Primary caregivers by gender
caregivers = pd.DataFrame({
    "gender": ["Women", "Men"],
    "population": [
        df_care.loc[10, 6],
        df_care.loc[10, 9]
    ]
})

# Calculate shares
caregivers["percentage"] = (
    caregivers["population"]
    / caregivers["population"].sum() * 100
)

caregivers
```

```python
# Create waffle grid
waffle = pd.DataFrame({"square": range(100)})

waffle["column"] = waffle["square"] % 20
waffle["row"] = waffle["square"] // 20

women_share = round(caregivers.loc[0, "percentage"])

waffle["gender"] = np.where(
    waffle["square"] < women_share, "Women", "Men"
)
```

```python
# Waffle chart
waffle_chart = (
    alt.Chart(waffle)
    .mark_square(size=650,opacity = 1)
    .encode(
        x=alt.X("column:O", axis=None),
        y=alt.Y("row:O", axis=None, sort="ascending"),
        color=alt.Color(
            "gender:N",
            scale=alt.Scale(
                domain=["Women", "Men"],
                range=["#8624f5", "#1fc3aa"]
            ),
            legend=None
        ),
        tooltip=["gender:N"]
    )
    .properties(
        width=850,
        height=230,
        title=alt.TitleParams(
            text="Who provides care in Mexico?",
            subtitle="Out of every 100 primary caregivers, 2022",
            anchor="middle",
            fontSize=20,
            subtitleFontSize=13
        )
    )
)

waffle_chart

```

```python
women_label = (
    alt.Chart(pd.DataFrame({
        "label": [f"{round(caregivers.loc[0, 'percentage'])} Women"]
    }))
    .mark_text(
        fontSize=30,
        fontWeight="bold",
        color="#8624f5"
    )
    .encode(text="label:N")
    .properties(width=400, height=55)
)

men_label = (
    alt.Chart(pd.DataFrame({
        "label": [f"{round(caregivers.loc[1, 'percentage'])} Men"]
    }))
    .mark_text(
        fontSize=30,
        fontWeight="bold",
        color="#1fc3aa"
    )
    .encode(text="label:N")
    .properties(width=400, height=55)
)

final_waffle = (
    alt.vconcat(
        waffle_chart,
        alt.hconcat(women_label, men_label, spacing=50),
        spacing=10
    )
    .configure_view(stroke=None)
)

final_waffle
```

```python
final_waffle.save("../images/care_numbers.png", scale_factor=2)
```

```python
#Next graph
# Read weekly caregiving hours
df_hours = pd.read_excel(
    file_path,
    sheet_name="Cuadro 3.15",
    header=None
)

df_hours.iloc[6:17, :]
```

```python
# Select caregiving categories and weekly hours
care_hours = df_hours.loc[
    [10, 12, 13, 14],
    [1, 13, 19]
].copy()

care_hours.columns = ["category", "women", "men"]

care_hours["category"] = [
    "All caregivers",
    "Disability or dependency",
    "Household members",
    "Other households"
]

care_hours
```

```python
care_long = care_hours.melt(
    id_vars="category",
    value_vars=["women", "men"],
    var_name="gender",
    value_name="hours"
)

care_long["gender"] = care_long["gender"].map({
    "women": "Women",
    "men": "Men"
})

care_long
```

```python
category_order = [
    "All caregivers",
    "Household members",
    "Disability or dependency",
    "Other households"
]

# Base chart
base = alt.Chart(care_long).encode(
    y=alt.Y(
        "category:N",
        sort=category_order,
        title=None,
        axis=alt.Axis(labelFontSize=13)
    ),
    x=alt.X(
        "hours:Q",
        title="Average weekly caregiving hours",
        scale=alt.Scale(domain=[0, 45]),
        axis=alt.Axis(grid=False)
    )
)

# Connecting lines
lines = (
    alt.Chart(care_hours)
    .mark_rule(color="#B0B0B0", strokeWidth=3)
    .encode(
        y=alt.Y("category:N", sort=category_order),
        x="men:Q",
        x2="women:Q"
    )
)

# Gender dots
dots = base.mark_circle(size=220).encode(
    color=alt.Color(
        "gender:N",
        scale=alt.Scale(
            domain=["Women", "Men"],
            range=["#8624f5", "#1fc3aa"]
        ),
        legend=alt.Legend(title=None, orient="bottom")
    ),
    tooltip=[
        "category:N",
        "gender:N",
        alt.Tooltip("hours:Q", title="Hours/week", format=".1f")
    ]
)

# Labels
labels = base.mark_text(
    dy=-17,
    fontSize=12,
    fontWeight="bold"
).encode(
    text=alt.Text("hours:Q", format=".1f"),
    color=alt.Color(
        "gender:N",
        scale=alt.Scale(
            domain=["Women", "Men"],
            range=["#8624f5", "#1fc3aa"]
        ),
        legend=None
    )
)

# Final chart
hours_chart = (
    (lines + dots + labels)
    .properties(
        width=600,
        height=260,
        title=alt.TitleParams(
            text="The unequal time burden of caregiving",
            subtitle="Average weekly hours among caregivers, Mexico, 2022",
            anchor="middle",
            fontSize=20,
            subtitleFontSize=13
        )
    )
    .configure_view(stroke=None)
)

hours_chart
```

```python
hours_chart.save("../images/time_dedicated_to_care.png", scale_factor=2)
```

```python
#Next graph
df_impact = pd.read_excel(
    file_path,
    sheet_name="Cuadro 3.11",
    header=None
)

df_impact.iloc[10:40, :]
```

```python
# Select economic and labor consequences
impact = df_impact.loc[
    [17, 22, 27, 32, 37],
    [7, 10]
].copy()

impact.columns = ["women", "men"]

impact["category"] = [
    "Unable to work outside home",
    "Had to stop working",
    "Reduced spending or borrowed money",
    "Lost promotion or career opportunities",
    "Reduced working hours or schedule difficulties"
]

impact = impact.reset_index(drop=True)

impact.round(1)
```

```python
impact_long = impact.melt(
    id_vars="category",
    value_vars=["women", "men"],
    var_name="gender",
    value_name="percentage"
)

impact_long["gender"] = impact_long["gender"].map({
    "women": "Women",
    "men": "Men"
})

impact_long
```

```python
# Category order and labels
category_order = impact["category"].tolist()

category_labels = {
    "Unable to work outside home": "Unable to work outside home",
    "Had to stop working": "Had to stop working",
    "Reduced spending or borrowed money": "Reduced spending or borrowed money",
    "Lost promotion or career opportunities": "Lost promotion or career opportunities",
    "Reduced working hours or schedule difficulties": "Reduced working hours or schedule difficulties"
}

impact_long["category_label"] = impact_long["category"].map(category_labels)

# Base chart
base = alt.Chart(impact_long).encode(
    y=alt.Y(
        "category_label:N",
        sort=category_order,
        title=None,
        axis=alt.Axis(
            labelFontSize=12,
            labelLimit=450,
            labelPadding=12
        )
    ),
    x=alt.X(
        "percentage:Q",
        title="Share of caregivers (%)",
        scale=alt.Scale(domain=[0, 35]),
        axis=alt.Axis(grid=False)
    )
)

# Gender dots
dots = base.mark_circle(
    size=220,
    opacity=1
).encode(
    color=alt.Color(
        "gender:N",
        scale=alt.Scale(
            domain=["Women", "Men"],
            range=["#8624f5", "#1fc3aa"]
        ),
        legend=alt.Legend(title=None, orient="bottom")
    ),
    tooltip=[
        "category:N",
        "gender:N",
        alt.Tooltip(
            "percentage:Q",
            format=".1f",
            title="Percentage"
        )
    ]
)

# Women's labels above the dots
women_labels = (
    base.transform_filter(
        alt.datum.gender == "Women"
    )
    .mark_text(
        dy=-18,
        fontSize=12,
        fontWeight="bold",
        color="#8624f5"
    )
    .encode(
        text=alt.Text("percentage:Q", format=".1f")
    )
)

# Men's labels below the dots
men_labels = (
    base.transform_filter(
        alt.datum.gender == "Men"
    )
    .mark_text(
        dy=18,
        fontSize=12,
        fontWeight="bold",
        color="#1fc3aa"
    )
    .encode(
        text=alt.Text("percentage:Q", format=".1f")
    )
)

# Final chart
impact_chart = (
    (dots + women_labels + men_labels)
    .properties(
        width=550,
        height=320,
        title=alt.TitleParams(
            text="Opportunity cost of caregiving",
            subtitle="Reported economic and labor impacts among caregivers, Mexico, 2022",
            anchor="middle",
            fontSize=20,
            subtitleFontSize=13
        )
    )
    .configure_view(stroke=None)
)

impact_chart
```
